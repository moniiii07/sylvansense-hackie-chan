```python
bbox = dict(min_lon=-62.2115, min_lat=-3.2915, max_lon=-62.2093, max_lat=-3.2893)
aoi = ee.Geometry.Rectangle([bbox["min_lon"], bbox["min_lat"], bbox["max_lon"], bbox["max_lat"]])

m = geemap.Map()
m.centerObject(aoi, zoom=15)
m.addLayer(aoi, {"color": "red"}, "AOI")
m
```

```python
s2 = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
        .filterBounds(aoi)
        .filterDate("2024-01-01", "2024-06-30")
        .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 15))
        .median())

m.addLayer(s2, {"bands": ["B4", "B3", "B2"], "min": 0, "max": 3000}, "True Color")
ndvi = s2.normalizedDifference(["B8", "B4"])
m.addLayer(ndvi, {"min": -0.2, "max": 0.9, "palette": ["red", "yellow", "green"]}, "NDVI")
m
```

```python
s1 = (ee.ImageCollection("COPERNICUS/S1_GRD")
        .filterBounds(aoi).filterDate("2024-01-01", "2024-06-30")
        .filter(ee.Filter.eq("instrumentMode", "IW"))
        .select(["VV", "VH"]).median())
m.addLayer(s1, {"bands": ["VV"], "min": -25, "max": 0}, "SAR VV")

chm = ee.Image("users/nlang/ETH_GlobalCanopyHeight_2020_10m_v1")
m.addLayer(chm, {"min": 0, "max": 40, "palette": ["white", "green", "darkgreen"]}, "Canopy Height")
m
```

```python
import sys
sys.path.append("src/ingestion")
from sentinel_fetch import fetch_sentinel_scene

scene = fetch_sentinel_scene(bbox)

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, (name, arr) in zip(axes.flat, scene.bands.items()):
    im = ax.imshow(arr, cmap="viridis")
    ax.set_title(f"{name}\nmin={arr.min():.2f} max={arr.max():.2f}")
    plt.colorbar(im, ax=ax, fraction=0.046)
plt.tight_layout()
```
